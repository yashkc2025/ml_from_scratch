```python
import numpy as np
from typing import List, Union

class LinearRegressor:
    type_arr = Union[List[List[float]], List[List[int]]]  # 2D input

    def __init__(self, X: type_arr, y: Union[List[float], List[int]]):
        self.X_arr = np.array(X)
        self.y_arr = np.array(y)
        self.coefficients = None  # b vector (including intercept)
        self.is_trained = False

    def train(self):
        X = self.X_arr
        y = self.y_arr

        # Add intercept term (column of 1s)
        ones = np.ones((X.shape[0], 1))
        X_b = np.hstack((ones, X))  # shape: (n_samples, n_features + 1)

        # Normal Equation: (X^T X)^-1 X^T y
        XTX = X_b.T @ X_b
        XTy = X_b.T @ y

        self.coefficients = np.linalg.pinv(XTX) @ XTy
        self.is_trained = True

    def get_weights(self):
        if not self.is_trained:
            raise RuntimeError("Model has not been trained yet.")
        return {
            "intercept": self.coefficients[0],
            "weights": self.coefficients[1:]
        }

    def predict(self, X_new: type_arr):
        if not self.is_trained:
            raise RuntimeError("Model has not been trained yet.")
        X_new = np.array(X_new)
        if X_new.ndim == 1:
            X_new = X_new.reshape(1, -1)  # single sample

        ones = np.ones((X_new.shape[0], 1))
        X_b_new = np.hstack((ones, X_new))
        return X_b_new @ self.coefficients
```

```python
X = [[1, 2], [2, 3], [4, 5]]
y = [3, 5, 9]

reg = LinearRegressor(X, y)
```

```python
reg.train()
```

```python
reg.get_weights()
```

```python
reg.predict([[5,7]])
```
